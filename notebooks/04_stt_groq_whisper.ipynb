# Layer 6 — Speech to Text

This checkpoint sends one recorded question to Groq Whisper and prints the transcript. Place a short spoken audio file at `data/sample_docs/sample_audio.wav`, or change `audio_path` below to your own supported file. The audio is uploaded to Groq when you run the transcription cell.

```python
from pathlib import Path
from time import perf_counter
import sys

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / 'src').is_dir():
    raise RuntimeError('Run this notebook from the ConversIQ project or notebooks directory.')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import SAMPLE_DOCS_DIR, STT_MODEL
from src.stt import transcribe_audio

audio_path = SAMPLE_DOCS_DIR / 'sample_audio.wav'
print(f'Audio path: {audio_path}')
print(f'STT model: {STT_MODEL}')
```

## Manual transcription check

Run this cell once after recording a clear spoken question. An empty transcript indicates that the model did not recognize speech in the sample.

```python
if not audio_path.is_file():
    raise FileNotFoundError(
        f'Place a spoken audio file at {audio_path}, or set audio_path to your file.'
    )

started = perf_counter()
transcript = transcribe_audio(audio_path)
elapsed = perf_counter() - started

print(f'Audio file: {audio_path.name}')
print(f'Transcription latency: {elapsed:.2f} s')
print(f'Transcript: {transcript or "<no speech recognized>"}')
assert isinstance(transcript, str)
```
