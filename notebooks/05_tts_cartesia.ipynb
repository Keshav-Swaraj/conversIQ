# Layer 7 — Cartesia text to speech

This checkpoint sends one short sentence to Cartesia and verifies the returned WAV. It uses the production `src.tts` API. Add `CARTESIA_API_KEY` to the project-root `.env` first; `CARTESIA_VOICE_ID` and `CARTESIA_MODEL` can override the defaults in `.env.example`. A run sends the demo text externally and may incur a small charge.

```python
from pathlib import Path
from time import perf_counter
import sys
import wave

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / 'src').is_dir():
    raise RuntimeError('Run this notebook from the ConversIQ project or notebooks directory.')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from IPython.display import Audio, display
from src.config import CARTESIA_API_KEY, CARTESIA_MODEL, CARTESIA_VOICE_ID
from src.tts import synthesize_speech

if not CARTESIA_API_KEY:
    raise RuntimeError('Add CARTESIA_API_KEY to the project-root .env before running this checkpoint.')
output_path = PROJECT_ROOT / 'cache' / 'tts' / 'layer7_cartesia_demo.wav'
demo_text = 'ConversIQ is ready to read your document aloud. You can ask a question at any time.'
print(f'Model: {CARTESIA_MODEL}; voice ID: {CARTESIA_VOICE_ID}')
print(f'Output WAV: {output_path}')
```

## Synthesize and inspect

Run this cell once when you want a real cloud TTS check. Confirm the WAV has non-empty audio, then listen for pronunciation, pacing, and naturalness.

```python
started = perf_counter()
audio_path = synthesize_speech(demo_text, output_path)
elapsed = perf_counter() - started
with wave.open(str(audio_path), 'rb') as wav_file:
    duration = wav_file.getnframes() / wav_file.getframerate()
    sample_rate = wav_file.getframerate()
assert duration > 0
print(f'Synthesis latency: {elapsed:.2f} s')
print(f'Audio duration: {duration:.2f} s; sample rate: {sample_rate} Hz')
display(Audio(filename=str(audio_path)))
```
